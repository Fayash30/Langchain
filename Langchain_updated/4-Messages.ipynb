```python
from langchain.chat_models import init_chat_model 
import os

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")

### Google Gemini
model = init_chat_model("google_genai:gemini-3.5-flash-lite")
```

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

message = [
    SystemMessage("You are a poetry expert and always respond in a Poetry way"),
    HumanMessage("Tell me About yourself")
]

response =  model.invoke(message)
response.content
```

### Tool Message

```python
from langchain.messages import ToolMessage

# After a model makes a tool call
# ( Here we demo manually creating the messages for ref)
ai_message = AIMessage(
    content=[],
    tool_calls = [{
        "name":"get_weather",
        "args":{"location":"Madurai"},
        "id":"call_2432"
    }]
)

# Execute tool and create result message
weather_result = "Sunny, 72F"
tool_message = ToolMessage(
    content=weather_result,
    tool_call_id = "call_2432" # Must match the Call id
)

#contiune conversation
messages = [
    HumanMessage("Whats the weather in Madurai"),
    ai_message, # Model's tool call
    tool_message # Tool exec result  
]

response = model.invole(messages)
```
