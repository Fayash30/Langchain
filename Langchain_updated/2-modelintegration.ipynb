### Model Integration with Open AI, Gemini and GROQ

```python
### Invoke the model

response = model.invoke("HI! How are you?")
response
```

### Streaming and Batch

### Streaming



```python
model = init_chat_model("google_genai:gemini-3.5-flash-lite")
model.stream("Write 200 words paragh on AI")
```

```python
for chunk in model.stream("Write 500 words paragh on AI"):
    print(chunk.text, end="|", flush= True)
```

### Batch

```python
responses = model.batch([
    "why do clock has 3 needle?",
    "How a fly can fly?",
    "what is capital of andaman nico"
])

for response in responses:
    print(response)
```

```python
responses = model.batch([
    "why do clock has 3 needle?",
    "How a fly can fly?",
    "what is capital of andaman nico"
], config= {'max_concurrency': 2})   # To limit the concurrency in larger batches
```
