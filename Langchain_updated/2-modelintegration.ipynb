### Model Integration with Open AI, Gemini and GROQ

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
from langchain.chat_models import init_chat_model 

### Google Gemini
model = init_chat_model("google_genai:gemini-3.5-flash-lite")
model

### For Chatgpt, just change the model name
```

```python
### Invoke the model

response = model.invoke("HI! How are you?")
response
```

```python
response.content
```

### Chat OpenAI

```python
from langchain_openai import ChatOpenAI

model = ChatOpenAI(model= "gpt-4.0")
response = model.invoke("How's life going?")
response


### No free tier
```

### ChatGoogleGenerative AI

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(model = "gemini-3.5-flash-lite")
response = model.invoke("Say Hello in Japanglish")
```

```python
response.content 
```
